```python
from sklearn.preprocessing import StandardScaler
cols = ['age','bmi','children']
scaler = StandardScaler()

df_cleaned[cols] = scaler.fit_transform(df_cleaned[cols])
 
```

```python
df_cleaned.head()
```

```python

from scipy.stats import pearsonr

# ----------------------------------
# Pearson Correlation Calculation
# ----------------------------------

# List of features to check against target
selected_features = [
    'age', 'bmi', 'children', 'is_female', 'is_smoker',
    'region_northwest', 'region_southeast', 'region_southwest',
    'bmi_catogery_underweight','bmi_catogery_normal_weight', 'bmi_catogery_over_weight', 'bmi_catogery_obese'
]

correlations = {
    feature: pearsonr(df_cleaned[feature], df_cleaned['charges'])[0]
    for feature in selected_features
}
correlation_df = pd.DataFrame(list(correlations.items()), columns=['Feature', 'Pearson Correlation'])
correlation_df.sort_values(by='Pearson Correlation', ascending=False)


```

```python
import pandas as pd
from scipy.stats import pearsonr

features = ['is_smoker', 'age', 'bmi_catogery_obese', 'bmi', 'region_southeast',
            'children', 'region_northwest', 'region_southwest',
            'bmi_catogery_underweight', 'is_female',
            'bmi_catogery_normal_weight', 'bmi_catogery_over_weight']

rows = []
for col in features:
    r, p = pearsonr(df_cleaned[col], df_cleaned['charges'])
    rows.append([col, r, p, "PASS (keep_feature)" if p < 0.05 else "FAIL (remove this feature)"])

result_df = pd.DataFrame(rows, columns=['Feature', 'r', 'p_value', 'Result'])
result_df = result_df.sort_values('p_value').reset_index(drop=True)
result_df
```
