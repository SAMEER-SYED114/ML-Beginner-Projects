K-Means Clustering

```python
import numpy as numpy
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
x,y=make_blobs(n_samples=500, centers=3, cluster_std=0.60
               , random_state=42)
```

```python
x
```

```python
df=pd.DataFrame(x,columns=['feature-1','feature-2'])
```

```python
df
```

```python
scaler=StandardScaler()
x_scaled=scaler.fit_transform(df)
```

```python
x_scaled
```

```python
inertia=[]
k_range=(1,11)
```

```python
for k in k_range:
    k_means=KMeans(n_clusters=k,random_state=42)
    k_means.fit(x_scaled)
    inertia.append(k_means.inertia_)
```

```python
inertia
```

```python
plt.plot(k_range,inertia,marker='o')
```

```python
k_means_final=KMeans(n_clusters=3,random_state=42)
```

```python
cluster_lable=k_means_final.fit_predict(x_scaled)
```

```python
cluster_lable
```

```python
df['cluster']=cluster_lable
```

```python
sns.scatterplot(
                x=df['feature-1'],
                y=df['feature-2'],
                hue=df['cluster'],
                palette='viridis'

)
```

DB SCAN

```python
from sklearn.datasets import make_moons
from sklearn.cluster import KMeans,DBSCAN
```

```python


from sklearn.datasets import make_moons

     

x, y = make_moons(n_samples=500, noise=0.05, random_state=42)

     

from sklearn.cluster import KMeans, DBSCAN
     

df = pd.DataFrame(x, columns=['Feature_1', 'Feature_2'])

     

scaler = StandardScaler()
x_scaled = scaler.fit_transform(df)
     

kmeans = KMeans(n_clusters=2, random_state=42)
kmeans_labels = kmeans.fit_predict(x_scaled)
     

df['kmeans_cluster'] = kmeans_labels

     

sns.scatterplot(x=df['Feature_1'], y=df['Feature_2'], hue=df['kmeans_cluster'], palette='tab10')

     
```

```python
X,Y=make_moons(n_samples=500,noise=0.05,random_state=42)
```

```python


dbscan = DBSCAN(eps=0.3, min_samples=5)
dbscan_labels = dbscan.fit_predict(x_scaled)
     

df['dbscan_cluster'] = dbscan_labels

     

sns.scatterplot(x=df['Feature_1'], y=df['Feature_2'], hue=df['dbscan_cluster'], palette='tab10')

     
```

PCA

```python
from sklearn.decomposition import PCA
```

```python
x,y=make_blobs(n_samples=500,n_features=5,centers=3,cluster_std=1.5,random_state=42)
```

```python
X_scaled=scaler.fit_transform(x)
```

```python
X_scaled
```

```python
pca=PCA(n_components=2)
x_pca=pca.fit_transform(X_scaled)
```

```python
x_pca
```

```python
df_pca=pd.DataFrame(x_pca,columns=['pc1','pc2'])
df_pca['lables']=y
```

```python
sns.scatterplot(data=df_pca,x='pc1',y='pc2',hue='lables',palette='Set2')
```
